## Stemming


```python
words = [
"playing","played","player","plays",
"studies","studying","studied",
"running","runner","runs",
"eating","eaten","eats",
"writing","written","writes",
"driving","driven","drives",
"quickly","happiness","hopeful",
"useful","useless","beautiful",
"cars","boxes","babies","cities","dogs",
"connection","connected","connecting",
"communication","communicating","organization"
]
```

### PorterStemmer


```python

from nltk.stem import PorterStemmer

stemming = PorterStemmer()

for word in words:
    print(word+'--->',stemming.stem(word))
```

```python
# DisAdvantage of stemming (bcs its change the meaning of word)
stemming.stem('Congratulation')
```

### RegExpStemmer 


```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)

reg_stemmer.stem('eating') 
```

### SnowbollStemmer

```python
from nltk.stem import SnowballStemmer

snow_stemming  = SnowballStemmer(language='english')

for word in words:
    print(word+'-->'+snow_stemming.stem(word))
```
