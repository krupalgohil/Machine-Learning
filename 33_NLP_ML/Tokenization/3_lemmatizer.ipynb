# Lemmatization 

## Wordnet Lemmatizer 



```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
words = [
"playing","played","player","plays",
"studies","studying","studied",
"running","runner","runs", 
"eating","eaten","eats",
"writing","written","writes",
"driving","driven","drives",
"quickly","happiness","hopeful",
"useful","useless","beautiful",
"cars","boxes","babies","cities","dogs",
"connection","connected","connecting",
"communication","communicating","organization"
]
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
# pos = v for virbs,n for noun,
for word in words:
    print(word+"-->"+lemmatizer.lemmatize(word,pos='n'))
```
