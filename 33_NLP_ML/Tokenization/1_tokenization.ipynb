# NLP Tokenization Tutorial

This notebook introduces **tokenization**, one of the first steps in Natural Language Processing (NLP).

Tokenization means breaking text into smaller pieces called **tokens** (words, sentences, or characters).

## 1. Install Required Library
We will use the **NLTK** library.

```python
# !pip install nltk
```

## 2. Import Libraries

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download('punkt')
```

## 3. Sentence Tokenization
Sentence tokenization splits text into sentences.

```python
text = "NLP is interesting. It is used in chatbots, search engines, and translation systems."

sentences = sent_tokenize(text)
sentences
```

## 4. Word Tokenization
Word tokenization splits sentences into words.

```python
words = word_tokenize(text)
words
```

## 5. Tokenization Example with User Text

```python
my_text = "I am learning Natural Language Processing to become a Data Scientist."

print("Sentence Tokens:")
print(sent_tokenize(my_text))

print("\nWord Tokens:")
print(word_tokenize(my_text))
```
