```python
speech = """
My dear young friends, today I would like to talk about dreams, goals, and the future of our nation. 
Dream is not that which you see while sleeping. Dream is something that does not let you sleep. 
Dreams transform into thoughts and thoughts result in action. 

When I was a young boy growing up in Rameswaram, I had dreams like many of you. 
I believed that with determination and hard work, any goal can be achieved. 
Education gives you knowledge, knowledge gives you creativity, creativity leads to thinking, 
thinking provides knowledge, and knowledge makes you great.

To succeed in your mission, you must have single-minded devotion to your goal. 
Difficulties in your life do not come to destroy you, but to help you realize your hidden potential and power. 
Let difficulties know that you too are difficult.

Great dreams of great dreamers are always transcended. 
If you want to shine like a sun, first burn like a sun. 
Man needs his difficulties because they are necessary to enjoy success.

All of us do not have equal talent, but all of us have an equal opportunity to develop our talents. 
The future of India depends on the youth of the country. 
Young minds are the most powerful resource on the earth, above the earth, and under the earth.

My message to the youth is to have courage to think differently, 
courage to invent, courage to travel the unexplored path, 
courage to discover the impossible, and courage to conquer the problems and succeed.

You have to dream before your dreams can come true. 
Excellence happens not by accident. It is a process. 
You must work hard, acquire knowledge, and never give up on your dreams.

Let us all work together to build a developed nation, a nation where every citizen 
has the opportunity to learn, grow, and contribute to society. 
Remember that the power of youth can transform the future of our country.

Dream, dream, dream. Dreams transform into thoughts and thoughts result in action.
"""
```

```python
import nltk 
from nltk.corpus import stopwords
sentences = nltk.sent_tokenize(speech)
```

```python
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
for i in range(len(sentences)):

    # word tokenization
    words = nltk.word_tokenize(sentences[i])

    # remove stopwords
    words = [word for word in words if word.lower() not in set(stopwords.words('english'))]

    # POS tagging
    pos_tag = nltk.pos_tag(words)

    print(pos_tag)
```
